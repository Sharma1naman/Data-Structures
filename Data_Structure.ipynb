**Question 2.** Explain the difference between mutable and immutable data types with examples.
- In Python, data types are categorized based on whether their values can be changed after the object is created.

 **Mutable data types** are those that **allow modification** of their contents. This means you can change, add, or remove elements without creating a new object. Common examples of mutable types include list, dict, and set.

 On the other hand, **immutable data types** are those that **do not allow any change** to their contents once they are created. If you try to change them, Python will raise an error or force you to create a new object. Examples of immutable types include int, float, str, tuple, and bool.



```python
# Example of Mutable data type
my_list = [1, 2, 3]
my_list[0] = 100
print(my_list)
```

```python
# Example of immutable data type
my_string = "Hello"

# Trying to modify a character will raise an error
# my_string[0] = "h"  → This will raise TypeError

# Correct way: create a new string
new_string = "h" + my_string[1:]
print(new_string)
```

**Question 3.** What are the main differences between lists and tuples in Python?
- In Python, **lists** and **tuples** are both used to store collections of items, but they have several key differences that affect how and when they should be used.

 The most significant difference is that **lists are mutable**, meaning their elements can be changed, added, or removed after the list is created. For example, you can append new items, remove existing ones, or update values in a list. This makes lists highly flexible and ideal for situations where the content of the collection is expected to change during the program’s execution.

 In contrast, **tuples are immutable**, which means that once a tuple is created, its elements cannot be modified. You cannot add, remove, or change items in a tuple. This immutability makes tuples more memory-efficient and slightly faster than lists. Tuples are often used when you want to ensure that a sequence of values remains constant throughout the program, such as fixed configurations or coordinates.

 Another difference is in **syntax:** lists are created using square brackets [], while tuples use parentheses (). Although both can store mixed data types, the immutability of tuples also makes them usable as keys in dictionaries, while lists cannot be used as dictionary keys because they are not hashable.

 In summary, if you need a flexible, editable collection, use a list. If you want a fixed, unchangeable collection that offers better performance and safety, go with a tuple.



**Question 4.** Describe how dictionaries store data?
- In Python, **dictionaries** store data as **key-value pairs**. Each item in a dictionary has two parts: a **key**, which is a unique identifier, and a **value**, which is the data associated with that key. You can think of a dictionary like a real-world dictionary where you look up a word (the key) to find its definition (the value).

 Dictionaries use a data structure called a **hash table** under the hood. When you add a key to a dictionary, Python uses a **hashing function** to convert that key into a unique index where the value is stored. This allows for **very fast lookups**—you can retrieve a value just by referencing its key, without searching through the entire collection.

 Keys in a dictionary must be **immutable** (such as strings, numbers, or tuples), while values can be of any data type, including lists, other dictionaries, or even functions. Dictionaries are **mutable**, meaning you can add, update, or remove key-value pairs after the dictionary is created.

 In short, dictionaries are powerful tools for managing data that needs to be labeled or accessed quickly using unique identifiers.

**Question 5.** Why might you use a set instead of a list in Python?
- You might choose to use a **set instead of a list** in Python when you need to store a **collection of unique items** and don't care about their order. Unlike lists, sets automatically **eliminate duplicates**, so they are ideal when you want to ensure that each element appears only once—for example, when collecting unique user inputs or filtering repeated values. Sets are also optimized for **fast membership tests**, meaning checking whether an item exists in a set (x in my_set) is generally faster than doing the same with a list. However, since sets are **unordered**, they do not maintain the order in which elements were added, and they don't support indexing or slicing like lists do. So, if uniqueness and quick lookup are more important than maintaining order or duplicates, a set is the better choice.



**Question 6.** What is a string in Python, and how is it different from a list?

**Question 7.** How do tuples ensure data integrity in Python?

This makes tuples ideal for storing data that should remain constant throughout the execution of a program, such as fixed settings, coordinates, or days of the week. **Question 8.** What is a hash table, and how does it relate to dictionaries in Python?
-  A **hash table** is a data structure that stores data in an **unordered collection of key-value pairs**, allowing for very **fast data retrieval**. The key idea behind a hash table is the use of a **hash function**—this function takes a key and converts it into a unique index (called a hash) that determines where the corresponding value is stored in memory. This enables quick access to values based on their keys, often in constant time, regardless of the size of the data set.

 In Python, the built-in dict **(dictionary)** type is implemented using a hash table under the hood. When you add a key-value pair to a dictionary, Python hashes the key to figure out where to store the value. Later, when you want to retrieve that value using the key, Python uses the same hash function to quickly locate it. This is what makes dictionary lookups (my_dict['key']) so efficient. However, because dictionary keys must be hashable, only **immutable data types** like strings, numbers, and tuples can be used as keys.

**Question 10.** Explain why strings are immutable in Python?
 - Strings are immutable in Python, meaning once a string is created, it cannot be changed. This design choice provides several benefits. Firstly, it enhances security and reliability by preventing accidental modification of data, making code more predictable and less prone to bugs. Secondly, immutability allows strings to be hashable, which is essential for using them as keys in dictionaries and storing them in sets—something mutable objects cannot do. Additionally, immutability enables Python to optimize performance through a technique called string interning, where identical strings are stored only once in memory. This saves space and improves efficiency. Lastly, immutability supports a functional programming style by reducing side effects, as functions that operate on strings return new strings rather than altering the original. To modify a string in Python, a new string must be created based on the desired changes.

**Queestion 11.** What advantages do dictionaries offer over lists for certain tasks?
 - Dictionaries in Python offer several advantages over lists for certain tasks, especially when it comes to storing and retrieving data efficiently. One of the main benefits is **fast lookups**—dictionaries provide average-case constant time complexity (O(1)) for retrieving values by key, whereas lists require a linear search (O(n)) if you're trying to find an element. This makes dictionaries ideal when you need to associate **unique keys with values** and access them quickly.

 Another advantage is that dictionaries offer **clearer and more meaningful organization** of data. For example, instead of accessing a user’s name by index in a list (user[0]), you can use a dictionary and access it by key (user["name"]), which improves readability and maintainability. Dictionaries also make it easier to **represent structured data**, such as JSON-like objects, and to **avoid duplication of keys**, helping maintain data integrity.

 Overall, dictionaries are preferred when you need fast, intuitive access to values through unique identifiers, while lists are better suited for ordered collections where the position of elements matters.

**Question 17.** Describe the time complexity of accessing elements in a dictionary.
 - The **time complexity of accessing elements in a dictionary** in Python is **O(1)** on average. This means that retrieving a value by its key is done in constant time, regardless of the size of the dictionary.

 Python dictionaries use a **hash table** internally. When you access an element using a key, Python calculates the hash of the key and directly looks up the corresponding value in memory—making it very fast.

 In rare cases (like hash collisions), the time complexity can degrade to **O(n)**, but Python’s hash algorithm is designed to **minimize collisions**, so this is uncommon.

**Question 19.** Why are dictionaries considered unordered, and how does that affect data retrieval?
 - Dictionaries were traditionally considered **unordered** because they didn’t maintain the order in which key-value pairs were added. However, starting from **Python 3.7**, dictionaries **preserve insertion order** by default. Despite this, they are still often conceptually treated as unordered collections because their primary purpose is **fast key-based access**, not ordering.

 This means that when you retrieve data from a dictionary, you're typically accessing it by **key**, not by position. So, even though the items may appear in the order you inserted them, you **shouldn't rely on that order for logic** unless you're explicitly working with Python 3.7+ and aware of this behavior.

 In summary, while modern dictionaries **do keep order**, they are still used for tasks where **quick lookups by key** matter more than the sequence of items. This approach allows dictionaries to provide **fast and efficient data retrieval**, typically in constant time (O(1)).










**Question 20.** Explain the difference between a list and a dictionary in terms of data retrieval.
 - The main difference between a **list** and a **dictionary** in terms of data retrieval lies in **how** you access the elements.

   - In a **list**, data is retrieved using **index positions** (e.g., my_list[0]). This means you access items based on their order in the list. If you don’t know the index of the item, you may need to **search through the list**, which can take time (O(n) in the worst case).

  - In a **dictionary**, data is retrieved using **keys** (e.g., my_dict["name"]). Dictionaries use a **hash table** under the hood, so accessing a value by its key is typically **very fast**—on average, it takes constant **time (O(1))** regardless of the size of the dictionary.

 In short, **lists are best** when you care about the **order or position** of items, while **dictionaries are better** when you need **quick access** to data through **unique identifiers (keys)**.
